```python
# Benötigte Bibliotheken
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# CSV-Datei laden
df = pd.read_csv(
    r"C:\Users\AZADAL01\Documents\scherer\games.csv",
    index_col=False
)

# -----------------------------
# Allgemeine Informationen
# -----------------------------

print("=" * 50)
print("ANZAHL ZEILEN UND SPALTEN")
print("=" * 50)

print("Zeilen:", df.shape[0])
print("Spalten:", df.shape[1])

print("\n")

print("=" * 50)
print("SPALTENNAMEN")
print("=" * 50)

for col in df.columns:
    print(col)

print("\n")

print("=" * 50)
print("ERSTE 5 ZEILEN")
print("=" * 50)

print(df.head())

print("\n")

print("=" * 50)
print("DATENTYPEN")
print("=" * 50)

print(df.info())

print("\n")

print("=" * 50)
print("FEHLENDE WERTE")
print("=" * 50)

print(df.isnull().sum())

print("\n")

print("=" * 50)
print("STATISTISCHE ÜBERSICHT")
print("=" * 50)

print(df.describe())

# -----------------------------
# Numerische Spalten finden
# -----------------------------

print("\n")
print("=" * 50)
print("NUMERISCHE SPALTEN")
print("=" * 50)

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    print(col)

# -----------------------------
# Histogramme für numerische Spalten
# -----------------------------

for col in numeric_columns:

    plt.figure(figsize=(8, 5))

    df[col].dropna().hist(bins=50)

    plt.title(f"Verteilung von {col}")
    plt.xlabel(col)
    plt.ylabel("Häufigkeit")

    plt.show()

# -----------------------------
# Korrelationen
# -----------------------------

if len(numeric_columns) > 1:

    print("\n")
    print("=" * 50)
    print("KORRELATIONSMATRIX")
    print("=" * 50)

    corr = df[numeric_columns].corr()

    print(corr)

print("\n")
print("=" * 50)
print("ANALYSE BEENDET")
print("=" * 50)
```
